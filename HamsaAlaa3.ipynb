#Task 10

**1. loc vs iloc**
`loc` selects rows and columns by label (index name). `iloc` selects by integer position, starting from 0.

**2. Inner join vs left join**
An inner join keeps only rows that have a match in both tables. A left join keeps all rows from the left table and fills unmatched columns from the right table with NaN.

**3. Duplicate keys in a lookup DataFrame**
If a key appears more than once in the lookup table, each matching row in the main table is joined with every duplicate. This creates extra rows and can inflate totals such as revenue.

**4. groupby vs pivot table**
Use `groupby` for simple summaries by one or more columns, especially with several named aggregations. Use `pivot_table` when you want one column's values spread across the columns, such as City as rows and Category as columns.

**5. Why handle cancelled orders and missing quantities first**
Cancelled orders are not real sales, and rows with a missing quantity cannot give a correct revenue. Including them would make the business totals wrong.

**6. Two benefits of Parquet**
1. It stores data by column, so you can read only the columns you need, which is faster.
2. It keeps the schema (column names and data types) and compresses the data, so files are smaller and load with the correct types.
